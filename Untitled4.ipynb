```python
import os
import zipfile
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

print("TensorFlow Version:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
```

```python
import shutil

DATASET_DIR = "/content/road_dataset"

if os.path.exists(DATASET_DIR):
    shutil.rmtree(DATASET_DIR)

os.makedirs(DATASET_DIR, exist_ok=True)

with zipfile.ZipFile("/content/dataset.zip", "r") as zip_ref:
    zip_ref.extractall(DATASET_DIR)

print("Dataset extracted successfully!")
```

```python
print("Dataset structure:\n")

for root, dirs, files in os.walk(DATASET_DIR):
    level = root.replace(DATASET_DIR, "").count(os.sep)

    if level <= 3:
        print("  " * level + os.path.basename(root) + "/")
```

```python
train_dir = None
test_dir = None

for root, dirs, files in os.walk(DATASET_DIR):

    folder_names = [d.lower() for d in dirs]

    if "train" in folder_names:
        train_dir = os.path.join(root, dirs[folder_names.index("train")])

    if "test" in folder_names:
        test_dir = os.path.join(root, dirs[folder_names.index("test")])

print("Train Path:", train_dir)
print("Test Path :", test_dir)
```

```python
print("Train Classes:")
print(os.listdir(train_dir))

print("\nTest Classes:")
print(os.listdir(test_dir))
```

```python
classes = sorted([
    folder for folder in os.listdir(train_dir)
    if os.path.isdir(os.path.join(train_dir, folder))
])

print("Detected Classes:")
print(classes)
```

```python

for cls in classes:

    train_path = os.path.join(train_dir, cls)
    test_path = os.path.join(test_dir, cls)

    train_count = len([
        f for f in os.listdir(train_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    test_count = len([
        f for f in os.listdir(test_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    print(f"\n{cls}")
    print("Train:", train_count)
    print("Test :", test_count)
```

```python
plt.figure(figsize=(12, 6))

plot_index = 1

for cls in classes:

    class_path = os.path.join(train_dir, cls)

    image_files = [
        f for f in os.listdir(class_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ][:5]

    for image_name in image_files:

        image_path = os.path.join(class_path, image_name)

        img = tf.keras.utils.load_img(
            image_path,
            target_size=(224, 224)
        )

        plt.subplot(len(classes), 5, plot_index)
        plt.imshow(img)
        plt.title(cls)
        plt.axis("off")

        plot_index += 1

plt.tight_layout()
plt.show()
```

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.20,
    rotation_range=15,
    width_shift_range=0.10,
    height_shift_range=0.10,
    zoom_range=0.15,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="training",
    shuffle=True
)

validation_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="validation",
    shuffle=False
)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    shuffle=False
)

print("\nClass Mapping:")
print(train_generator.class_indices)
```

```python
model = models.Sequential([

    layers.Input(shape=(224, 224, 3)),

    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(256, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),

    layers.Dense(1, activation="sigmoid")
])

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("CNN model compiled successfully!")
```

```python
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=20
)

print("\nCNN TRAINING COMPLETED!")
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("CNN Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.subplot(1, 2, 2)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("CNN Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
test_loss, test_accuracy = model.evaluate(test_generator)

print("\n==============================")
print("ROAD DAMAGE DETECTION RESULTS")
print("==============================")
print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")
print("==============================")
```

```python
test_generator.reset()

predictions = model.predict(test_generator)

y_pred = (predictions >= 0.5).astype(int).flatten()
y_true = test_generator.classes

class_names = list(test_generator.class_indices.keys())

print("Classes:", class_names)
```

```python
print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names
    )
)
```

```python
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Road Damage Detection - Confusion Matrix")

plt.show()
```

```python
test_generator.reset()

images, labels = next(test_generator)

predictions = model.predict(images)

plt.figure(figsize=(14, 10))

for i in range(min(12, len(images))):

    actual = class_names[int(labels[i])]
    predicted = class_names[int(predictions[i] >= 0.5)]

    plt.subplot(3, 4, i + 1)

    plt.imshow(images[i])

    plt.title(
        f"Actual: {actual}\nPredicted: {predicted}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from google.colab import files
from tensorflow.keras.utils import load_img, img_to_array

uploaded = files.upload()

for filename in uploaded.keys():

    img = load_img(
        filename,
        target_size=IMG_SIZE
    )

    img_array = img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    prediction = model.predict(img_array)[0][0]

    if prediction >= 0.5:
        result = "POTHOLE / DAMAGED ROAD"
        confidence = prediction * 100
    else:
        result = "PLAIN / NORMAL ROAD"
        confidence = (1 - prediction) * 100

    plt.figure(figsize=(7, 7))
    plt.imshow(img)
    plt.title(
        f"{result}\nConfidence: {confidence:.2f}%"
    )
    plt.axis("off")
    plt.show()

    print("Prediction:", result)
    print(f"Confidence: {confidence:.2f}%")
```
